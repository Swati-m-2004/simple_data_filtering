```python


# -----------------------------
# 1. Create student records
# -----------------------------
students = [
    {"id": 101, "name": "Aarav", "age": 20, "department": "CSE", "marks": 88},
    {"id": 102, "name": "Ananya", "age": 21, "department": "ECE", "marks": 76},
    {"id": 103, "name": "Rahul", "age": 22, "department": "ISE", "marks": 91},
    {"id": 104, "name": "Sneha", "age": 20, "department": "CSE", "marks": 69},
    {"id": 105, "name": "Vivek", "age": 23, "department": "ME", "marks": 72},
    {"id": 106, "name": "Priya", "age": 21, "department": "CSE", "marks": 95},
    {"id": 107, "name": "Kiran", "age": 22, "department": "ECE", "marks": 81},
    {"id": 108, "name": "Megha", "age": 20, "department": "ISE", "marks": 74},
    {"id": 109, "name": "Rohan", "age": 21, "department": "CSE", "marks": 84},
    {"id": 110, "name": "Pooja", "age": 23, "department": "CIVIL", "marks": 67},
    {"id": 111, "name": "Akash", "age": 20, "department": "CSE", "marks": 79},
    {"id": 112, "name": "Divya", "age": 22, "department": "ECE", "marks": 89},
]

# -----------------------------
# 2. Display all records
# -----------------------------
def display_students(records, title):
    print("\n" + "=" * 75)
    print(title)
    print("=" * 75)

    if not records:
        print("No students found.")
        return

    print(f"{'ID':<6}{'Name':<12}{'Age':<6}{'Department':<12}{'Marks':<6}")
    print("-" * 75)

    for student in records:
        print(
            f"{student['id']:<6}"
            f"{student['name']:<12}"
            f"{student['age']:<6}"
            f"{student['department']:<12}"
            f"{student['marks']:<6}"
        )

# -----------------------------
# 3. Filter by marks
# Condition: marks >= 80
# -----------------------------
high_scorers = [
    student for student in students
    if student["marks"] >= 80
]

# -----------------------------
# 4. Filter by age
# Condition: age <= 21
# -----------------------------
young_students = [
    student for student in students
    if student["age"] <= 21
]

# -----------------------------
# 5. Filter by department
# Condition: department == "CSE"
# -----------------------------
cse_students = [
    student for student in students
    if student["department"] == "CSE"
]

# -----------------------------
# 6. Multiple-condition filtering
# Conditions:
# - Department is CSE
# - Marks are at least 80
# -----------------------------
cse_high_scorers = [
    student for student in students
    if student["department"] == "CSE" and student["marks"] >= 80
]

# -----------------------------
# 7. Another multiple-condition filter
# Conditions:
# - Age <= 21
# - Marks >= 75
# -----------------------------
young_high_scorers = [
    student for student in students
    if student["age"] <= 21 and student["marks"] >= 75
]

# -----------------------------
# 8. Display results
# -----------------------------
display_students(students, "ALL STUDENT RECORDS")
display_students(high_scorers, "FILTER 1: STUDENTS WITH MARKS >= 80")
display_students(young_students, "FILTER 2: STUDENTS WITH AGE <= 21")
display_students(cse_students, "FILTER 3: STUDENTS FROM CSE DEPARTMENT")
display_students(
    cse_high_scorers,
    "FILTER 4: CSE STUDENTS WITH MARKS >= 80"
)
display_students(
    young_high_scorers,
    "FILTER 5: AGE <= 21 AND MARKS >= 75"
)

# -----------------------------
# 9. Summary
# -----------------------------
print("\n" + "=" * 75)
print("FILTERING SUMMARY")
print("=" * 75)
print(f"Total students              : {len(students)}")
print(f"Marks >= 80                 : {len(high_scorers)}")
print(f"Age <= 21                   : {len(young_students)}")
print(f"CSE department               : {len(cse_students)}")
print(f"CSE and marks >= 80          : {len(cse_high_scorers)}")
print(f"Age <= 21 and marks >= 75    : {len(young_high_scorers)}")

# -----------------------------
# 10. Optional Pandas example
# -----------------------------
# This section shows how the same idea can be applied to a DataFrame.
# Uncomment the lines below if Pandas is installed.

# import pandas as pd
# df = pd.DataFrame(students)
#
# pandas_result = df[
#     (df["department"] == "CSE") &
#     (df["marks"] >= 80)
# ]
#
# print("\nPandas DataFrame Filter:")
# print(pandas_result)
```
